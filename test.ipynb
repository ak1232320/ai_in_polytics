```python
import numpy as np

# List of possible outcomes (numpy array recommended)
outcomes = np.array(['Red', 'Black', 'Green'])

# Corresponding probabilities (must sum to 1.0)
probabilities = [0.45, 0.45, 0.10] # e.g., 45%, 45%, 10%

# Select a single element (size=1) based on probabilities
selected_element = np.random.choice(outcomes, size=1, p=probabilities)

print(f"The selected element is: {selected_element}")
```
